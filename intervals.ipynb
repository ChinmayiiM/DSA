Summary ranges

```python
def summaryRanges(self, nums: List[int]) -> List[str]:
        result = []
        start = 0
        for i in range(1, len(nums) + 1):
            if i == len(nums) or nums[i] != nums[i - 1] + 1:
                if start == i - 1:
                    result.append(str(nums[start]))
                else:
                    result.append(str(nums[start]) + "->" + str(nums[i - 1]))
                start = i
        return result
```
