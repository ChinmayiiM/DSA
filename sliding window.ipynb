Substring with concatenation of all word


```python
def findSubstring(self, s: str, words: List[str]) -> List[int]:
        answer = []
        word_len = len(words[0])
        total_len = len(words)
        total_len = word_len * total_len
        word_count = {}
        for word in words:
            if word in word_count:
                word_count[word] += 1
            else:
                word_count[word] = 1
        for i in range(len(s) - total_len + 1):
            seen = {}
            j = i
            while j < i + total_len:
                word = s[j:j+ word_len]
                if word not in word_count:
                    break
                if word in seen:
                    seen[word] += 1
                else:
                    seen[word] =1
                if seen[word] > word_count[word]:
                    break
                j += word_len
            if j == i + total_len:
                answer.append(i)
        return answer
```

Maximum window substring

```python
def minWindow(self, s, t):
        need = {}
        for ch in t:
            if ch in need:
                need[ch] += 1
            else:
                need[ch] = 1
        left = 0
        ans = ""
        for right in range(len(s)):
            if s[right] in need:
                need[s[right]] -= 1
            while True:
                ok = True
                for ch in need:
                    if need[ch] > 0:
                        ok = False
                        break
                if not ok:
                    break
                if ans == "" or right - left + 1 < len(ans):
                    ans = s[left:right + 1]
                if s[left] in need:
                    need[s[left]] += 1
                left += 1
```

Longest substring without Repeating Characters

```python
def lengthOfLongestSubstring(self, s):
        seen = set()
        left = 0
        max_len = 0
        for right in range(len(s)):
            while s[right] in seen:
                seen.remove(s[left])
                left += 1
            seen.add(s[right])
            max_len = max(max_len, right-left + 1)
        return max_len
```
