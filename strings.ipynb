Reverse  string

```python
def reverseString(self,s):
    s[:] = s[::-1]
```

First Non repeating characters


```python
def first_non_repeating(s):
    count =  {}
    for ch in s:
        if ch in count:
            count[ch] += 1
        else:
            count[ch] = 1
    for ch in s:
        if count[ch] == 1
            return ch
    return none 
```

IS Subsequence

```python
def isSubsequence(self, s, t):
    i = 0
    j = 0
    while i < len(s) and j < len(t):
        if s[i] != t[j]:
            i += 1
        j += 1
    return i == len(s)
```
