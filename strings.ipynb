Reverse  string

```python
def reverseString(self,s):
    s[:] = s[::-1]
```

First Non repeating characters


```python
def first_non_repeating(s):
    count =  {}
    for ch in s:
        if ch in count:
            count[ch] += 1
        else:
            count[ch] = 1
    for ch in s:
        if count[ch] == 1
            return ch
    return none 
```
