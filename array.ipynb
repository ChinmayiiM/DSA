Intersection of Two arrays

```python
def intersection(self,nums1, nums2):
    result = []
    for num in nums1:
        if num in nums2 and num not in result:
            result.append(num)
    return result

```

Merge Two sorted arrays

```python
def merge(self,nums1, m, nums2, n):
    merge = nums1[:m] + nums2[:n]
    merge.sort()
    nums1[:] = merge
```

Missing number in array

```python
def ismissing(self,nums):
    n = len(nums)
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)
    missing = expected_sum - actual_sum
    return missing
```

Anagram

```python
def isAnagram(self,s,t):
    if len(s) != len(t):
        return False
    return sorted(s) == sorted(t)
```

Palindrome Number

```python
def isPalindrome(self,x):
    if str(x) == str(x)[::-1]:
        return True
    else:
        return False
```

Rearrange array

```python
def rearrangeArray(self,nums):
    positive = []
    negative = []
    for num in nums:
        if num > 0:
            positive.append(num)
        else:
            negative.append(num)
    result = []
    for i in range(len(positive)):
        result.append(positive[i])
        if i < len(negative):
            result.append(negative[i])
    return result
```

Move zeroes to end

```python
def moveZeroes(self,nums):
    result = []
    zero_count = 0
    for num in nums:
        if num == 0:
            zero_count += 1
        else:
            result.append(num)  
    for i in range(zero_count):
        result.append(0)
    nums[:] = result
```

```python
def sortColors(self,nums):
    low = 0
    mid = 0
    high = len(nums) - 1

    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
```
