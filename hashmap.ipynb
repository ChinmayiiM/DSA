Valid Anagram


```python
def isAnagram(self, s, t):
        if len(s) != len(t):
            return False
        return sorted(s) == sorted(t)
```

Group Anagram

```python
def groupAnagrams(self, strs):
        freq ={}
        for word in strs:
            key = "".join(sorted(word))
            if key not in freq:
                freq[key] = []
            freq[key].append(word)
        return list(freq.values())
```

Isomorphic

```python
def isisomorphic(self,s,t):
    map1= {}
    map2= {}
    for i in range(len(s)):
        if s[i] in map1 and map1[s[i]] != t[i]:
            return False
        if t[i]  in map2 and map2[t[i]] != s[i]:
            return False
        map1[s[i]] = t[i]
        map2[t[i]] = s[i]
    return True
```

Happy Number

```python
def isHappy(self,n):
    seen = set()
    while n != 1:
        if n in seen:
            return False
        seen.add(n)
        total = 0
        while n > 0:
            digit = n % 10
            total += digit * digit
            n = n//10
        n = total
    return True
```
