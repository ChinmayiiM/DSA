Valid Anagram


```python
def isAnagram(self, s, t):
        if len(s) != len(t):
            return False
        return sorted(s) == sorted(t)
```

Group Anagram

```python
def groupAnagrams(self, strs):
        freq ={}
        for word in strs:
            key = "".join(sorted(word))
            if key not in freq:
                freq[key] = []
            freq[key].append(word)
        return list(freq.values())
```
