Detect cycle

```python
def has_cycle(head)
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False
```

Search Element


```python
def search(head,value):
    current = head
    while current:
        if current.data == value:
            return True
        current = current.next
    return False
```
