Reverse stack


```python
def reverse_stack(stack):
    temp = []
    while stack:
        temp.append(stack.pop())
    return temp
```

postfix Expression

```python
def evaluate_postfix(expression):
    stack = []
    for ch in expression:
        if ch.isdigit():
            stack.append(int(ch))
        else:
            b = stack.pop()
            a = stack.pop()
            if ch == "+":
                stack.append(a+b)
            elif ch == "-":
                stack.append(a-b)
            elif ch == "*":
                stack.append(a*b)
            elif ch == "/"
                stack.append(a//b)
    reverse stack.pop()
```
